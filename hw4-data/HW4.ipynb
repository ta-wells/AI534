```python
class BasicPerceptron:
    def __init__(self, input_dim, learning_rate=0.1):
        self.weights = np.zeros(input_dim)  # Initialize weights to 0
        self.bias = 0  # Initialize bias to 0
        self.learning_rate = learning_rate
    
    def predict(self, x):
        # Simple step function: If the weighted sum is positive, predict 1, else 0
        linear_output = np.dot(x, self.weights) + self.bias
        return np.where(linear_output >= 0, 1, 0)
    
    def train(self, X, y, dev,y_dev,epochs=10):
        best_dev = 1
        for epoch in range(epochs):
            
            for i in range(len(X)):
                # Make a prediction for the current sample
                prediction = self.predict(X[i])
                # Calculate error
                error = y[i] - prediction
                # Update weights and bias if there is an error (Perceptron learning rule)
                self.weights += self.learning_rate * error * X[i]
                self.bias += self.learning_rate * error
            prediction = self.predict(dev)
            test =1- accuracy_score(y_dev,prediction)
            if test<best_dev:
                best_dev = test
                best_epoch = epoch+1
            print("Epoch {0:1d}: Dev Error: {1:.2%}".format(epoch+1,test))
        return best_dev,best_epoch
```

```python
class AveragedPerception:
    def __init__(self, input_dim, learning_rate=.1, averaging_factor=0.045):
        self.weights = np.zeros(input_dim)  # Initialize weights to 0
        self.bias = 0  # Initialize bias to 0
        self.avg_weights = np.zeros(input_dim)  # Initialize averaged weights to 0
        self.avg_bias = 0  # Initialize averaged bias to 0
        self.learning_rate = learning_rate
        self.averaging_factor = averaging_factor  # How much to weigh the current weights in the average

    def predict(self, X):
        # Compute the linear combination for the entire batch
        linear_output = np.dot(X, self.avg_weights) + self.avg_bias  # Use averaged weights for prediction
        # Apply step function to each example in the batch
        return np.where(linear_output >= 0, 1, 0)
    
    def train(self, X, y, dev,y_dev,epochs=10):
        best_dev = 1
        for epoch in range(epochs):
            for i in range(len(X)):
                # Make a prediction for the current sample
                prediction = self.predict(X[i].reshape(1, -1))  # Reshape for single sample
                prediction = prediction[0]  # Extract prediction value
                # Calculate error
                error = y[i] - prediction
                # Update weights and bias if there is an error (Perceptron learning rule)
                self.weights += self.learning_rate * error * X[i]
                self.bias += self.learning_rate * error
                
                # Update the averaged weights and bias
                self.avg_weights = (1 - self.averaging_factor) * self.avg_weights + self.averaging_factor * self.weights
                self.avg_bias = (1 - self.averaging_factor) * self.avg_bias + self.averaging_factor * self.bias
            prediction = self.predict(dev)
            test =1- accuracy_score(y_dev,prediction)
            if test<best_dev:
                best_dev = test
                best_epoch = epoch+1
            print("Epoch {0:1d}: Dev Error: {1:.2%}".format(epoch+1,test))

        return best_dev,best_epoch,prediction
```

```python
input_dim = len(average_list[0])  # Dimension of sentence embeddings
perceptron_av = AveragedPerception(input_dim,averaging_factor=.00000000001)

# Train the perceptron on the sentence embeddings
print("Starting Averaged Perceptron")
best_dev_av,best_epoch_av,prediction_1 = perceptron_av.train(average_list, y_train_kNN,average_list_dev,y_dev_kNN, epochs=10)

print("\nBest Dev Error: {0:.2%} at epoch {1:1d}".format(best_dev_av,best_epoch_av))
```
